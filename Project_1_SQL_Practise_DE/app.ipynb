```python
# iloc and loc
# iloc is works as numpy index and it will exclude the right side range
# iloc takes integer values as input, range works like left is inclusive and right is exclusive  e.g: data.iloc[0:2,0:2]

# loc, we need to give the column names
# loc function, it will include both right and left side of the range (left is inclusive and right is inclusive).
# between column names range operators works as well.

# data.loc[0:2,['Row ID', 'Customer Name']]
# data.iloc[0:2,       0:2]
#           |           |
#           |           |
#          <>       this is representing the columns from start:end


#  this is representing the rows from [start:end]


# iloc is handle with numerics
# loc is handle with both numerics and string(columns)
```
