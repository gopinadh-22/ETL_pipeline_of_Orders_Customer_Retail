```python
#example for class methods

class Employee:
    company_name = 'Tech_corp' #--> two class attributes
    Employee_count = 0

    #instance methods
    #constructor
    def __init__(self,name,salary):
        self.name = name
        self.salary = salary
        Employee.Employee_count += 1

    #class methods
    #decorator
    @classmethod
    def change_company_name(cls,new_company_name):
        cls.company_name = new_company_name
        print('company name changed to:',{cls.company_name})
    
    #classmethod
    @classmethod
    def get_employee_count(cls):
        return cls.Employee_count

    #instance method
    def display_info(self):
        print(f'Employee:{self.name}, company:{Employee.company_name}')
```

```python
emp1 = Employee('rajesh',50000)
```

```python
emp2 = Employee('anjali',60000)
```

```python
# static method

class Mathoperator():
    @staticmethod
    def add(a,b):
        return a + b
    @staticmethod
    def mul(a,b):
        return a * b
    @staticmethod
    def is_even(number):
        return (number % 2 == 0)

var1 = Mathoperator.add(2,3)
var2 = Mathoperator.mul(4,4)
var3 = Mathoperator.is_even(4)
print(var1,var2,var3)
```

```python
# Question: create a bankaccount class in python

# 1. create a bankaccount class
# 2. that demonstrates instance methods, class methods, and static methods using banking system
# 3. define class attribute bank_name = 'abc' and total_accounts = 0
# 4. create a constructor that accepts account holder and balance. 
#     store them as instance attributes and increment total_accounts whenever a new account is created.
# 5. create an instance method deposit(amount) that adds money to the account balance
# 6. create an instance method display_balance() that displays the account holder's name and current balance.
# 7. create a class method change_bank_name(new_name) that updates the bank name for all acoounts.
# 8. create a class method get_total_accounts() that return the total number of accounts created.
# 9. create a static method is_valid_amount(amount) that returns True if the amount is greater than zero; otherwise, return False.
# 10. create two bank account objects and test all three types of methods.
```

```python
class BankAccount():
    #class_attributes
    bank_name = 'ABC'
    total_accounts = 0

    #constructor creation and instance attributes
    def __init__(self,account_holder,balance):
        self.account_holder = account_holder
        self.balance = balance
        BankAccount.total_accounts += 1
    
    #instance method
    def deposit(self,amount=0):
        self.balance += amount
        print(f'update_balance:{self.balance}')
    
    #instance method
    def display_balance(self):
        print(f'account_holder:{self.account_holder}')
        print(f'current_balance:{self.balance}')

    def display_info(self):
            print(f'Employee:{self.account_holder}, company:{BankAccount.bank_name}, Amount_balance:{self.balance}')

    @classmethod
    def change_bank_name(cls,new_name):
        cls.bank_name = new_name
        print('bank name changed to:',{cls.bank_name})

    @classmethod
    def get_total_accounts(cls):
        print(f'total_account:{cls.total_accounts}')

    @staticmethod
    def is_valid_amount(amount):
        if(amount > 0):
            return True
        else:
            return False
```
